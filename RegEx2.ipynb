```python
import re

email= input("Enter Email:")
email_pat= r'^[A-Za-z0-9._%+-]+@[A-Za-z0-9]+\.[a-z]{2,}$'

if re.match(email_pat,email):
    print("Email is Valid")
else:
    print("Email is not Valid")
```

```python
phone= input("Enter Phone  No:")
phone_pat= r'^[6-9]\d{9}$'

if re.match(phone_pat,phone):
    print(phone+" Phone Number is Valid")
else:
    print(phone+" Phone Number is not Valid")
```

```python
date= input("Enter Date(DD/MM/YYYY):")
date_pat= r'^(0[1-9]|[12][0-9]|3[0-1])/(0[1-9]|1[0-2])/\d{4}$'

if re.match(date_pat,date):
    print(date+" Date is Valid")
else:
    print(date+" Date is not Valid")
```

```python
url= input("Enter URL:")
url_pat= r'^https?:\/\/(www\.)?[A-Za-z0-9.-]+\.[A-Za-z]{2,}$'

if re.match(url_pat,url):
    print(url+" is Valid")
else:
    print(url+" is not Valid")
    
```

```python
hashtag = input("Enter Sentence:")
hashtag_pat= r'^#[A-Za-z0-9_]+$'
if re.match(hashtag_pat,hashtag):
    print(hashtag+" is Valid Hashtag")
else:
    print(hashtag+" is not Valid Hashtag")
```

```python
adhar= input("Enter Adhaar No:")
adhar_pat= r'^\d{12}$'
if re.match(adhar_pat,adhar):
    print(adhar+" Adhaar Number is Valid")
else:
    print(adhar+" Adhaar Number is not Valid")
```

```python
!pip install nltk spacy
!python -m spacy download en_core_web_sm
import nltk
import spacy

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger_eng')

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk import pos_tag

text = "The students are studying machine learning and they are learning quickly."

tokens = word_tokenize(text)

stop_words = set(stopwords.words('english'))
filtered = [word for word in tokens if word.lower() not in stop_words]

lemmatizer = WordNetLemmatizer()
lemmas = [lemmatizer.lemmatize(word) for word in filtered]

pos = pos_tag(tokens)

print("NLTK")
print("Tokens:", tokens)
print("After Stop-word Removal:", filtered)
print("Lemmatization:", lemmas)
print("POS Tagging:", pos)

nlp = spacy.load("en_core_web_sm")
doc = nlp(text)

spacy_tokens = [token.text for token in doc]
spacy_filtered = [token.text for token in doc if not token.is_stop]
spacy_lemmas = [token.lemma_ for token in doc]
spacy_pos = [(token.text, token.pos_) for token in doc]

print("\nspaCy")
print("Tokens:", spacy_tokens)
print("After Stop-word Removal:", spacy_filtered)
print("Lemmatization:", spacy_lemmas)
print("POS Tagging:", spacy_pos)

print("\nMorphological Components")
for token in doc:
    print(token.text, "Root:", token.lemma_, "Suffix:", token.suffix_)

print("\nAmbiguity Analysis")
sentence1 = "I saw a bat."
sentence2 = "The chicken is ready to eat."

print(sentence1, "- Lexical ambiguity")
print(sentence2, "- Syntactic ambiguity")
```
