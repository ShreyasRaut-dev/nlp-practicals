```python
import numpy as np
import pandas as pd

from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer


documents = [
    "machine learning is useful",
    "machine learning is powerful",
    "deep learning uses neural networks",
    "machine learning uses data",
    "data science uses machine learning",
    "python is useful for data science",
    "deep learning is a part of machine learning",
    "python is used for machine learning",
    "data science is useful",
    "neural networks are used in deep learning"
]


words = sorted(set(" ".join(documents).lower().split()))

one_hot = []

for doc in documents:
    doc_words = set(doc.lower().split())
    row = [1 if word in doc_words else 0 for word in words]
    one_hot.append(row)

one_hot_df = pd.DataFrame(one_hot, columns=words)

print("\nONE-HOT ENCODING")
print(one_hot_df)


bow_vectorizer = CountVectorizer()

bow = bow_vectorizer.fit_transform(documents)

print("\nBAG OF WORDS")
print(bow_vectorizer.get_feature_names_out())
print(bow.toarray())


dtm = pd.DataFrame(
    bow.toarray(),
    columns=bow_vectorizer.get_feature_names_out()
)

print("\nDOCUMENT-TERM MATRIX")
print(dtm)


tfidf_vectorizer = TfidfVectorizer()

tfidf = tfidf_vectorizer.fit_transform(documents)

tfidf_df = pd.DataFrame(
    tfidf.toarray(),
    columns=tfidf_vectorizer.get_feature_names_out()
)

print("\nTF-IDF")
print(tfidf_df.round(3))



def sparsity(matrix):
    total = matrix.shape[0] * matrix.shape[1]
    zero_values = total - np.count_nonzero(matrix)
    return (zero_values / total) * 100


print("\nCOMPARISON")

print("One-Hot Shape:", one_hot_df.shape)
print("BoW Shape:", bow.shape)
print("DTM Shape:", dtm.shape)
print("TF-IDF Shape:", tfidf.shape)

print("\nSparsity:")
print("One-Hot:", round(sparsity(np.array(one_hot)), 2), "%")
print("BoW:", round(sparsity(bow.toarray()), 2), "%")
print("DTM:", round(sparsity(dtm.values), 2), "%")
print("TF-IDF:", round(sparsity(tfidf.toarray()), 2), "%")


print("\nFEATURE VECTOR OF FIRST DOCUMENT")

print("Document:", documents[0])

print("\nBoW Vector:")
print(bow.toarray()[0])

print("\nTF-IDF Vector:")
print(tfidf.toarray()[0].round(3))
```
