```python

document = ["I am Shreyas",
            "I love NLP",
            "Artificial Intelligence is a field of Computer Science that mimic the human behaviour and train the agents",
            "Machine Learning algos learn from data",
            "Data Science is a field that extracts useful insights from raw information",
            "Classification and Regression are the algorithms of Machine Learning"
           ]
for i in document:
    print(i)
```

```python
from sklearn.preprocessing import OneHotEncoder
words=[["AI"],["NLP"],["ML"],["CC"],["BM"]]
encoder = OneHotEncoder(sparse_output=False)
Encoded_data = encoder.fit_transform(words)
print("\nOne Hot Encoding...!\n")
print("Vacabulary=",encoder.categories_[0])
print(Encoded_data)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
Vectorizer=CountVectorizer()
BOW=Vectorizer.fit_transform(document)
print("\nBag of Words...!\n")
print("Vacabulary=",Vectorizer.get_feature_names_out())
print(BOW.toarray())
```

```python
import pandas as pd

DTM_df=pd.DataFrame(
     BOW.toarray(),
     columns = Vectorizer.get_feature_names_out(),
     index = [f"Document {i}" for i in range(1,len(document)+1)]
 )

print("\nDocument Term Matrix...!\n")
print(DTM_df)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
Vectorizer2 = TfidfVectorizer()
TFIDF_Data=Vectorizer2.fit_transform(document)

TFIDF_df = pd.DataFrame(
 TFIDF_Data.toarray(),
 columns = Vectorizer2.get_feature_names_out(),
 index = [ f"Document {i}" for i in range(1,len(document)+1)]
)
print("\nTF-IDF...!\n")
print(TFIDF_df)
```

```python
print("\n Sparse Representation VS Dense Representation...!\n")
print("Sparse:")
print(BOW)
print("Dense:")
print(BOW.toarray())
```

```python
import matplotlib.pyplot as plt

feature_names = Vectorizer.get_feature_names_out()
document_vector = BOW.toarray()[0]

plt.figure(figsize=(10, 5))

plt.bar(feature_names, document_vector)

plt.xlabel("Words")
plt.ylabel("Frequency")
plt.title("BoW Feature Vector - Document 1")

plt.xticks(rotation=45)

plt.show()


```
