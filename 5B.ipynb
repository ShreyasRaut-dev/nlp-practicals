```python
import spacy
from spacy import displacy

nlp = spacy.load("en_core_web_sm")

sentences = [
    "The boy eats an apple.",
    "Ravi plays cricket in the park.",
    "The girl reads a book.",
    "Students study Python carefully."
]

for sentence in sentences:
    doc = nlp(sentence)

    print("\nSentence:", sentence)
    print("\nWord\tPOS\tDependency\tHead")

    for token in doc:
        print(token.text, "\t", token.pos_, "\t",
              token.dep_, "\t", token.head.text)

    print("\nSubjects:")
    for token in doc:
        if token.dep_ in ["nsubj", "nsubjpass"]:
            print(token.text)

    print("Objects:")
    for token in doc:
        if token.dep_ in ["dobj", "obj", "pobj"]:
            print(token.text)

    print("Modifiers:")
    for token in doc:
        if token.dep_ in ["amod", "advmod", "det", "advcl"]:
            print(token.text)

    displacy.serve(doc, style="dep")
```
