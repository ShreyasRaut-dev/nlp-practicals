```python
text="My name is Shreyas, currentely Pursuing Btech in Artificial Intelligence and Data Science. "
print(text)
```

```python
from nltk.tokenize import word_tokenize
tokens = word_tokenize(text)
for i in tokens:
    print(i)
```

```python
from nltk.corpus import stopwords
stop_words = set(stopwords.words("english"))
print(stop_words)
```

```python
filtered_data=[]
for i in tokens:
    if i.lower() not in stop_words:
        filtered_data.append(i)
```

```python
print(filtered_data)
```
