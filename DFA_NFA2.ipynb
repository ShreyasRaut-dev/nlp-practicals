```python
binary=input("Enter Binary(0-1) Number:")
state="q0"
for i in binary:
    if state=="q0":
        if i == "0":
            state= "q1"
        elif i == "1":
            state = "q0"

    elif state == "q1":
        if i == "0":
            state = "q1"       
        elif i == "1":
            state = "q2"

    elif state == "q2" :
        if i == "0":
            state = "q0"
        elif i == "1":
            state = "q1"

if state=="q2":
    print("Accepted...!")
else:
    print("Rejected...!")
```
